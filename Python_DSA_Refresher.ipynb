**Question 1:** 
Write a program that asks the user for a number of days and prints the number of seconds.

```python
#ask user to input number of days
days = int(input("Enter the number of days: "))
#calculate the number of seconds in the given number of days
seconds_in_a_day = 24 * 60 * 60
total_seconds = days * seconds_in_a_day
#display the result
print(f"There are {total_seconds:,} seconds in {days} day(s).")
```

**Question 2:** 
Ask the user for the radius and calculate the volume of a sphere using (4/3)πr^3.

```python
import math
#ask user to input the radius of the sphere
radius = float(input("Enter the radius of the sphere: "))
#calculate the volume of the sphere
volume = (4/3) * math.pi * radius**3
#display the result
print(f"The volume of the sphere with radius {radius} is {volume:.2f}.")
```

**Question 3:** 
Using functions, compute the area and perimeter of a square; ask for side length and display results.

```python
# to calculate the area of a square
def calculate_area(side):
    return side ** 2

# to calculate the perimeter of a square
def calculate_perimeter(side):
    return side * 4

# Main 
def main():
    # Ask user for side length
    side_length = float(input("Enter the side length of the square: "))
    
    # Compute area and perimeter by calling the functions
    area = calculate_area(side_length)
    perimeter = calculate_perimeter(side_length)
    
    # Display  results
    print(f"\nFor a square with a side length of {side_length}:")
    print(f"Area: {area:.2f}")
    print(f"Perimeter: {perimeter:.2f}")

# Run program
if __name__ == "__main__":
    main()
```

**Question 4:** 
Write a program using functions that determines whether an input character is uppercase or lowercase.

```python
def check_character_case(char):
    """Determines the case of a given character."""
    if char.isupper():
        return "Uppercase"
    elif char.islower():
        return "Lowercase"
    else:
        return "Neither (it's a number, symbol, or space)"

def main():
    # Ask  user for input
    user_input = input("Enter a single character: ")
    
    # Ensure the user entered something
    if len(user_input) > 0:
        
        character = user_input[0]
        
        # Call the function to get the result
        case_result = check_character_case(character)
        
        # Display the result
        print(f"The character '{character}' is **{case_result}**.")
    else:
        print("You didn't enter any characters!")

# Run the program
if __name__ == "__main__":
    main()
```

**Question 5:** 
Translate the given pseudocode (x=0, y=20; repeat: y-=4; x+=2/y; until y<6; display x) into Python.

```python
# set initial values for x and y
x=0
y=20
# repeat the loop until y is less than 6
while True:
    y=y-4
    x=x+2/y
# stop the loop if y is less than 6
    if y<6:
        break
    # display the current values of x 
    print(x)
```

**Question 6:** 
Use a loop to input 5 values into an array, then calculate and display their average.

```python
# Initialize an empty list to store the values
values = []

print("Please enter 5 values:")

# Use a for loop to continually prompt the user 5 times
for i in range(5):
    # Convert input to float to handle both integers and decimals
    val = float(input(f"Enter value {i + 1}: "))
    values.append(val)

# Calculate the average
total_sum = sum(values)
average = total_sum / len(values)

# Display the array and the final average
print(f"\nYour values: {values}")
print(f"The average of the values is: {average:.2f}")
```
